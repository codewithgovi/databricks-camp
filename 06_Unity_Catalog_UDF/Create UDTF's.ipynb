### UDTF's: User Defined Table Functions.
_User-Defined Table Functions (UDTFs) in Databricks are a type of user-defined function that allows you to register functions which return entire tables_

```python
%sql
CREATE OR REPLACE FUNCTION databricksgovi.bronze.udtf(param1 INT)
RETURNS TABLE
RETURN
(
    SELECT * FROM databricksgovi.silver.sales_enr WHERE total_amount > param1
)
```

```python
%sql
SELECT * FROM databricksgovi.bronze.udtf(150)
```
