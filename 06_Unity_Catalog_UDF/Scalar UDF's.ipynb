### SCALAR FUNCTIONS:
_A scalar Function is a function that takes one or more inputs and returns a single value_.
- "Scalar" = One Row In --> One Value Out

**Now I want to create the Scalar Functions for Bronze volumme**

### **SQL**

```python
%sql
CREATE OR REPLACE FUNCTION databricksgovi.bronze.sql_scalar_func(par1 INT)
RETURNS INT
LANGUAGE SQL
RETURN par1 * 10
```

```python
%sql
SELECT total_amount, databricksgovi.bronze.sql_scalar_func(total_amount) FROM databricksgovi.silver.sales_enr;
```

**Create scalar function on PYTHON**

```python
%sql
CREATE OR REPLACE FUNCTION databricksgovi.bronze.python_scalar_func(par1 INT)
RETURNS INT
LANGUAGE PYTHON
AS
$$
    return par1 * 10
$$
```

```python
%sql
SELECT total_amount, databricksgovi.bronze.python_scalar_func(total_amount) FROM databricksgovi.silver.sales_enr;
```

**Create a customizable scalar function in python**

```python
%sql
CREATE OR REPLACE FUNCTION python_scalar_func_cust(par1 INT)
RETURNS INT
LANGUAGE PYTHON
AS
$$
    import requests
    response = requests.get(f"rohanveer{par1}").json()
    response_str = str(response)
    return response_str
$$
```
