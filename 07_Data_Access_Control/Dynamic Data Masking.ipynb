### Dynamic Data Masking:
_Dynamic data masking is a security technique that hides sensitive data at query time without modifying the actual data stored in the table_,
Users with resticted permissions see a masked/obfuscated version of data, while authorized users see the full value


**Here We Should Apply Masking for Customer Table From Silver Layer**

```python
%sql
SELECT * FROM databricksgovi.silver.customer_enr;
```

**Basically We should display the customer email ID's**
_Now We will Apply MASKING for email column_

**CREATING A MASK FUNCTION**

```python
%sql
CREATE OR REPLACE FUNCTION databricksgovi.silver.dynamic_mask(p_email STRING)
RETURNS STRING
RETURN CASE WHEN is_account_group_member('developers') THEN p_email ELSE '*****' END;
```

**Applying the MASK**

```python
%sql
ALTER TABLE databricksgovi.silver.customer_enr
ALTER COLUMN email SET MASK databricksgovi.silver.dynamic_mask
```

```python
%sql
SELECT * FROM databricksgovi.silver.customer_enr;
```
