### **AUTOLOADER QUERY**

```python
# AUTOLOADER INITIALIZATION
df = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format","csv")\
    .option("cloudFiles.schemaLocation", "/Volumes/databricksgovi/bronze/autoload_vol/destination2/checkpoint/")\
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")\
    .load("/Volumes/databricksgovi/bronze/autoload_vol/raw/")
```

```python
df.writeStream.format("delta")\
        .outputMode("append")\
        .option("checkpointLocation", "/Volumes/databricksgovi/bronze/autoload_vol/destination2/checkpoint/")\
        .trigger(once=True)\
        .option("mergeSchema", "true")\
        .start("/Volumes/databricksgovi/bronze/autoload_vol/destination2/data/")
```
