### CTAS: Create Table As Select
_This is a Sql Command used to CREATE a new table and immediately populate it with the results of a SELECT query._

```python
%sql
CREATE TABLE databricksgovi.silver.ctas_tbl
AS
SELECT * FROM databricksgovi.silver.sales_enr
```

```python
%sql
SELECT * FROM databricksgovi.silver.ctas_tbl
```

```python
%sql
DESCRIBE HISTORY databricksgovi.silver.ctas_tbl
```

### DEEP CLONE
_Deep Clone creates a fully physical copy of a Delta table including all the data files and entire transaction history_

```python
%sql
CREATE TABLE databricksgovi.silver.deep_clone_tbl
CLONE databricksgovi.silver.sales_enr
```

```python
%sql
SELECT * FROM databricksgovi.silver.deep_clone_tbl
```

```python
%sql
DESCRIBE HISTORY databricksgovi.silver.deep_clone_tbl
```

### SHALLOW CLONE
_A Shallow clone creates a new delta table that copies only the metadata references the original tables data files, instead of masking a full physical copy_

```python
%sql
CREATE TABLE databricksgovi.silver.shallow_clone_tbl
SHALLOW CLONE databricksgovi.silver.sales_enr
```
