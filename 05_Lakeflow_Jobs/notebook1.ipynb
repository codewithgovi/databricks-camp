## **LakeFlow Jobs Implementation**

**Creating the Volume**

```python
%sql
-- CREATE VOLUME databricksgovi.bronze.jobvolume
```

**Transfer the stores info from bronze volume to job volume**

```python
df = spark.read.format('csv')\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .load('/Volumes/databricksgovi/bronze/bronze_volume/stores/')
```

```python
df.write.format('delta')\
    .mode('overwrite')\
    .save('/Volumes/databricksgovi/bronze/jobvolume/raw/stores')
```

**Transfer the sales info from bronze volume to job volume**

```python
df_sales = spark.read.format('csv')\
    .option('header', 'true')\
    .option('inferSchema', 'true')\
    .load('/Volumes/databricksgovi/bronze/bronze_volume/sales/')
```

```python
df_sales.write.format('delta')\
    .mode('overwrite')\
    .save('/Volumes/databricksgovi/bronze/jobvolume/raw/sales')
```
