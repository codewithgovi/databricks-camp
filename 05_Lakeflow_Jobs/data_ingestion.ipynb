**Here we are ingesting the data from bronze_vol to target_vol**

Note: If suppose, we have more folders we can't copy paste code.

```python
dbutils.widgets.text("folder_name", "")
dbutils.widgets.text("parent_folder_name","")
```

```python
folder_name = dbutils.widgets.get("folder_name")
parent_folder_name = dbutils.widgets.get("parent_folder_name")
```

We need to use f Strings here to pass the variable

```python
df = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
    .load(f"/Volumes/databricksgovi/bronze/bronze_volume/{folder_name}/")
```

```python
df.write.format("delta")\
        .mode("overwrite")\
        .save(f"/Volumes/databricksgovi/bronze/target_vol/{parent_folder_name}/{folder_name}/")
```
