## **Volumes:** 
A Volume is a governed folder in unity catalog used to store files with managed permissions, tracking, and metadata 

*-----Creating a Volume*

```python
%sql
CREATE VOLUME databricksgovi.bronze.bronze_volume
```

```python
%sql
SELECT * FROM csv.`/Volumes/databricksgovi/bronze/bronze_volume/sales/fact_sales.csv`
```

_That's how we govern the data by creating the volume under unity catalog_

## **DBUTILS**
Databricks Utilities are tools that can be used in Python, R, and Scala notebooks to perform various tasks efficiently.
- It's like built-in helper library that provides convinent commands to interact with the databricks environment.

**Key Functions:**
File and Object Storage || Secret Management || Widgets in Notebooks || Job task Utilities || Library Installation ||Notebook execution 

```python
dbutils.help()
```

```python
dbutils.fs.help()
```

**Ex 1: Now i want to check how many files are present in sales folder**

```python
dbutils.fs.ls("/Volumes/databricksgovi/bronze/bronze_volume/sales/")
```

- _fs means File System_
- _ls means list of files which contains inside a given folder_

**Ex 2:Create a customer folder using dbutils under bronze layer**

```python
dbutils.fs.mkdirs("/Volumes/databricksgovi/bronze/bronze_volume/customers")
```

**Ex 3: Now copy all the files from sales to customers**

_Here "cp" is used to COPY the resources_

```python
dbutils.fs.cp("/Volumes/databricksgovi/bronze/bronze_volume/sales/", "/Volumes/databricksgovi/bronze/bronze_volume/customers/", True)
```

**Ex 4: Now remove the files from customer folder**

_Here "rm" is used to REMOVE the resources_

```python
dbutils.fs.rm("/Volumes/databricksgovi/bronze/bronze_volume/customers/fact_sales.csv")
```

**Ex 5: Create a csv file under customer folder and provide any data**

_Here "put" is used to CREATE the resources_

```python
dbutils.fs.put("/Volumes/databricksgovi/bronze/bronze_volume/customers/test.csv","Hi Govi", True)
```

**Ex 6: Create a txt, python file under customer folder and provide any data**

```python
dbutils.fs.put("/Volumes/databricksgovi/bronze/bronze_volume/customers/test.py","print(Hi Govi)", True)
dbutils.fs.put("/Volumes/databricksgovi/bronze/bronze_volume/customers/test.txt","Hey Govi",True)
```

**Ex 7: Now find how many files inside customers folder**

```python
dbutils.fs.ls("/Volumes/databricksgovi/bronze/bronze_volume/customers/")
```

**Ex 8: Fetch only list of file names from the output**

```python
files = dbutils.fs.ls("/Volumes/databricksgovi/bronze/bronze_volume/customers/")
files
```

```python
file_names = [f.name for f in files]
file_names
```

### **WIDGETS:**
- Widgets in Databricks notebooks let you create input controls (dropdowns, text boxes, multiselect, comboboxes)

**Ex 9: Create a widget and store any value and reuse it**

```python
dbutils.widgets.text("parameter1","")
```

```python
para1_value = dbutils.widgets.get("parameter1")
para1_value
```

```python
para1_value
```

**Ex 10: Remove the files recursively inside the customer folder**

```python
for i in file_names:
    dbutils.fs.rm(f"/Volumes/databricksgovi/bronze/bronze_volume/customers/{i}")
```
